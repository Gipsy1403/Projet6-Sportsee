```python
from mistralai import Mistral
import os
from dotenv import load_dotenv
load_dotenv()



with Mistral(
    api_key=os.getenv("MISTRAL_API_KEY", ""),
) as mistral:

    res = mistral.chat.complete(model="mistral-small-latest", messages=[
        {
            "content": "tu es un coach sportif spécialisé en nutrition sportive, en programmation d'entraînement et en bien-être.",
            "role": "system",
        },  
       {
            "content": "peux tu me donner un régime adpaté pour une personne de 63 kilos qui fait de la course à pied 2fois par semaine soit 5 km par séance ?",
            "role": "user",
        },
    ], 
    temperature= 0.3,
    max_tokens= 150,
    stream=False)

    # Handle response
    print(res)

```
